### 6-1 데이터 훑어보기

```python
import pandas as pd
import numpy as np

data = {
    'name': ['김민수', '이지영', '박준호', '최서연', '정도윤'],
    'age': [25, 30, 28, 22, 35],
    'city': ['서울', '부산', '인천', '서울', '대전'],
    'score': [90, 85, 95, 80, np.nan]   #결측치를 설명하기 위해 np.nan 상수도 하나 같이 사용됨
}
df = pd.DataFrame(data)

# 데이터 측정
print(df.describe())

# 모든 데이터 타입 측정
df.describe(include='all')

# 수치형 데이터 측정
df.describe(include=[np.number])

# 범주형 데이터 측정
df.describe(include=['object'])

# 특정 열 분석
df['score'].describe()
```

### 6-2 DataFrame 열 선택 및 **조작**

```python
import pandas as pd

data = {
    "이름": ["김철수", "이영희", "박민수", "최지훈", "정소희"],
    "학년": [1, 2, 3, 4, 2],
    "학점": [4.2, 3.8, 4.5, 3.9, 3.5],
    "학과": ["컴퓨터공학", "경영학", "전자공학", "의학", "심리학"],
    "동아리": ["프로그래밍", "독서토론", "로봇공학", "봉사활동", "음악감상"]
}
df = pd.DataFrame(data)

# 특정 열 선택
# 하나의 열 df['이름']
# 다수의 열
df[['이름','학과','학점']]
#다수의 열을 객체로
selected_col = df[['이름','학과','학점']]
selected_col

# DataFrame 열순서 변경
reordered_cols=df[['학과','이름','학년','학점','동아리']]
reordered_cols
# DataFrame 열이름 변경
df.rename(columns={'이름':'학생명', '학점':'평점', '동아리':'과외활동'})
df
```

### 6-3 DataFrame 행 선택

```python
import pandas as pd

data = {
    "이름": ["김철수", "이영희", "박민수", "최지훈", "정소희"],
    "학년": [1, 2, 3, 4, 2],
    "학점": [4.2, 3.8, 4.5, 3.9, 3.5],
    "학과": ["컴퓨터공학", "경영학", "전자공학", "의학", "심리학"],
    "동아리": ["프로그래밍", "독서토론", "로봇공학", "봉사활동", "음악감상"]
}

#DataFrame 인덱스 지정
df = pd.DataFrame(data)
df=df.set_index('이름')
df
#DataFrame 행 선택, set_index 이름으로 안하면 그냥 일반적인 인덱스 0, 1, 3 같은 인덱스 넘버를 지정해주면 된다. df.loc[1]
#df.loc[0]
df.loc['김철수']
df.loc[['김철수','이영희']]

#파이썬에서 배웠던 것처럼 인덱스 슬라이싱해서 행선택
df.loc['김철수':'박민수']

#DataFrame 위치 기반 행 선택, 몇번째 행을 가져와 (index 위치 선택하기 iloc)
df.iloc[1]
df.iloc[[2,3]]
df.iloc[1:3]

#DataFrame 조건 기반 행 선택
df.loc[[True,False,True,False,False]]
print(df['학점'])
df.loc[df['학점']>=4.0]
df[df['학년'] ==2]
df[(df['학년']==2) & (df['학점']>=3.7)]
```

### 6-4 DataFrame 정렬

```python
import pandas as pd

data = {
    "이름": ["김철수", "이영희", "박민수", "최지훈", "정소희"],
    "학년": [1, 2, 3, 4, 2],
    "학점": [4.2, 3.8, 4.5, 3.9, 3.5],
    "학과": ["컴퓨터공학", "경영학", "전자공학", "의학", "심리학"],
    "동아리": ["프로그래밍", "독서토론", "로봇공학", "봉사활동", "음악감상"]
}

#DataFrame 인덱스 지정
df=pd.DataFrame(data).set_index('이름')
df

#DataFrame 단일 열 기본값 오름차순, 내림차순 정렬
df.sort_values(by='학점') #오름차순
df.sort_values(by='학점', ascending=False) #내림차순

#DataFrame 복수 열 오름차순, 내림차순 정렬
#학년은 오름차순인데 그 중에서 학점을 내림차순으로 정렬하고 싶을 때
df.sort_values(by=['학년','학점'], ascending=[True,False])


#DataFrame 인덱스 기준 오른차순, 내림차순 정렬, 인덱스가 이름이니까 이름 내림차순으로 됨
df.sort_index(ascending=False)
```
